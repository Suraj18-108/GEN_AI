#2. Loading GPT-2

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

model.eval()

print("GPT-2 model loaded successfully!")
```

#3. Building Text-Generation Function

```python
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0,
                  repetition_penalty=1.0, max_new_tokens=80):

    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=repetition_penalty,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_text = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return generated_text

do_sample=True
```

#4. Original Prompt :

```python
original_prompt = "The train arrived at the station late at night, but nobody stepped out."

print(original_prompt)
```

# Experiment 01 : Temperature [1.0 -> 0.3]


```python

exp1 = generate_text(
    original_prompt,
    temperature=0.3,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0,
    max_new_tokens=80
)

print(exp1)
```

# Experiment 02 : top-k [50 -> 10]

```python
exp2 = generate_text(
    original_prompt,
    temperature=1.0,
    top_k=10,
    top_p=1.0,
    repetition_penalty=1.0,
    max_new_tokens=80
)

print(exp2)
```

# Experiment 3 : top-p [1.0 -> 0.5]

```python
# experiment 3 top-p (top_p = 1.0 → 0.5)

exp3 = generate_text(
    original_prompt,
    temperature=1.0,
    top_k=50,
    top_p=0.5,
    repetition_penalty=1.0,
    max_new_tokens=80
)

print(exp3)
```

#Experiment 4 : repetition_penalty [1.0 -> 1.5]

```python


exp4 = generate_text(
    original_prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.5,
    max_new_tokens=80
)

print(exp4)
```

#Experiment 5 : max_new_tokens [80 -> 50]


```python

exp5 = generate_text(
    original_prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0,
    max_new_tokens=50
)

print(exp5)
```

#6. Custom Prompt :

```python
custom_prompt = "Our company has just launched a smart backpack that can charge electronic devices while you travel."

print(custom_prompt)
```

```python
custom_1 = generate_text(
    custom_prompt,
    temperature=0.5,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0,
    max_new_tokens=80
)

custom_2 = generate_text(
    custom_prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0,
    max_new_tokens=80
)

custom_3 = generate_text(
    custom_prompt,
    temperature=1.3,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0,
    max_new_tokens=80
)

print("VERSION 1:\n", custom_1)
print("\nVERSION 2:\n", custom_2)
print("\nVERSION 3:\n", custom_3)
```

# OBSERVATION :

1. Temperature (0.3): Kept the text focused and straightforward. The model just picked safe, obvious words, so it didn't get very creative/random.

2. Top-k (10): Limited the model to only its top 10 word choices each time, making the response a lot more predictable.

3. Top-p (0.5): Cut down on random words by sticking to the top 50% most likely options, which kept things clear and on point.

4. Repetition Penalty (1.5): Stopped the model from repeating itself, so it used different words instead of repeating the same phrases.

5. Max New Tokens (50): Just limited how long the answer could be. It made the response short without changing the actual style.
