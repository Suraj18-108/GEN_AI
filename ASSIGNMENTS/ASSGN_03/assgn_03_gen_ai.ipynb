#2. Load GPT-2

```python
generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

#3. Prompt Generation

```python
prompt = "The doctor entered the hospital and"

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

print(result[0]["generated_text"])

prompts = {
    "Doctor": "The doctor entered the hospital and",
    "Nurse": "The nurse entered the hospital and",
    "Business Executive": "The business executive arrived at the office and",
    "Scientist": "The scientist entered the laboratory and",
    "Police Officer": "The police officer arrived at the station and",
    "Criminal": "The criminal entered the building and"
}
```

#4. Prompt Checking

```python
for profession, prompt in prompts.items():
    print(f"{profession}: {prompt}")
```

#5. Sampling

```python
results = {}

for profession, prompt in prompts.items():
    outputs = generator(
        prompt,
        max_new_tokens=50,
        num_return_sequences=5,
        do_sample=True,
        temperature=0.8,
        top_p=0.9
    )

    results[profession] = [output["generated_text"] for output in outputs]
```

#6. Display The Result

```python
for profession, outputs in results.items():
    print("\n" + "=" * 60)
    print(profession)
    print("=" * 60)

    for i, output in enumerate(outputs, 1):
        print(f"\nOutput {i}:")
        print(output)
```

#7. Analyze the generated outputs

```python
for profession, outputs in results.items():
    print("\n" + "=" * 70)
    print(profession.upper())
    print("=" * 70)

    for i, output in enumerate(outputs, 1):
        print(f"\n{i}. {output}")
```
